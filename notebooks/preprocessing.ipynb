```python
import sys
sys.path.append('..')

import pandas as pd
import joblib
from sklearn.model_selection import GroupShuffleSplit

from src.preprocessing import load_raw, clean_data, split_columns
```

```python
raw = load_raw('../Data/diabetic_data.csv')
X, y, groups = clean_data(raw)

print('Rows before: ', len(raw), ' | after: ', len(X))
print('Columns before:', raw.shape[1], ' | feature after: ', X.shape[1])
print('Positive class rate: {: .2%}'.format(y.mean()))
```

```python
X['diag_1'].value_counts()
```

```python
print('Missing values left:', X.isna().sum().sum())
num_cols, cat_cols = split_columns(X)
print('Numeric:', num_cols)
print('Categorical:', cat_cols)
```

```python
gss = GroupShuffleSplit(n_splits=1, test_size=0.2, random_state=42)
train_idx, test_idx = next(gss.split(X, y, groups=groups))

X_train, X_test = X.iloc[train_idx].reset_index(drop=True), X.iloc[test_idx].reset_index(drop=True)
y_train, y_test = y.iloc[train_idx].reset_index(drop=True), y.iloc[test_idx].reset_index(drop=True)
g_train, g_test = groups.iloc[train_idx].reset_index(drop=True), groups.iloc[test_idx].reset_index(drop=True)

overlap = set(g_train) & set(g_test)
print('Train:', X_train.shape, '| Test:', X_test.shape)
print('Patients in both sets:', len(overlap))
print('Positive rate train {:.2%} | test {:.2%}'.format(y_train.mean(), y_test.mean()))
```

```python
joblib.dump({
    'X_train': X_train, 'X_test': X_test,
    'y_train': y_train, 'y_test': y_test,
    'g_train': g_train, 'g_test': g_test,
}, '../data/processed/split.joblib')
print('Saved.')
```
